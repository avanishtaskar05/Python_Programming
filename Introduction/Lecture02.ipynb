```python
# Conditional Statements in Python
# # if - else statement
marks = float(input('Enter the marks'))
if marks >= 90:
    print('Grade is O')
elif marks >=80:
    print('Grade is A+')
elif marks >= 70:
    print('Grade is A')
elif marks >= 60:
    print('Grade is B+')
elif marks >= 50:
    print('Grade is A')
else:
    print('Fail')
```

```python
availability = 16
booking = int(input('Enter the no of seat you want to book'))
new_availabilty = availability - booking
if booking <= availability:
    print('Booking is done')
else: 
    print('Booking is not  done')
print(f'Seats availability : {new_availabilty}')
```
