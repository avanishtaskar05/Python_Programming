```python
# Conditional Statements in Python
# # if - else statement
marks = float(input('Enter the marks'))
if marks >= 90:
    print('Grade is O')
elif marks >=80:
    print('Graded is A+')
elif marks >= 70:
    print('Grade is A')
elif marks >= 60:
    print('Grade is B+')
elif marks >= 50:
    print('Grade is A')
else:
    print('Fail')
```
