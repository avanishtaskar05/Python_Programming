```python
# WAP for asking user to enter a number and print the factorial of that number
def factorial():
    num = int(input("Enter a number for factorial: "))
    factorial = 1
    for i in range(1,num+1):
        factorial *= i
    print(f"The factorial of {num} is {factorial}")
factorial() 
```

```python
# WAP for asking user to enter a number and print the factorial of that number (with return function)
def factorial():
    num = int(input("Enter a number for factorial: "))
    factorial = 1
    for i in range(1,num+1):
        factorial *= i
    return factorial

ans = factorial() 
print(ans)
```

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
choice = int(input("Enter 1 for addition, 2 for subtraction, 3 for multiplication, 4 for division: "))

def sum(num1, num2):
    return num1 + num2

def sub(num1, num2):
    return num1 - num2

def mul(num1, num2):
    return num1 * num2

def div(num1, num2):
    return num1 / num2

if choice == 1:
    ans = sum(num1, num2)
    print(ans)

elif choice == 2:
    ans = sub(num1, num2)
    print(ans)

elif choice == 3:
    ans = mul(num1, num2)
    print(ans)

elif choice == 4:
    ans = div(num1, num2)
    print(ans)

else:
    print("Invalid choice")
```
