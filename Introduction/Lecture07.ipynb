```python
'''
class and object:
class: blueprint or template 
object: instance of a class
Note:
1. any function when defined inside a class is called "Methods of Respective Class"
2. any variable when defined inside a class is called "Attribute of Respective Class"
'''
```

```python
#class bank
class bank:
    # Constructor of a class    
    def __init__(self, name):
        self.name = name

    def name(self): # Method of a class
        name = self.name

    # object of the class
sbi = bank('SBI')
```

```python
class vehicle:

    def __init__(self, wheel, color, engine):
        self.wheel = wheel
        self.color = color
        self.engine = engine

    def sound(self):
        print("beep")

tata = vehicle(4, 'Red', 'Diesel')
audi = vehicle(4, 'Black', 'Petrol')
suzuki = vehicle(2, 'Blue', 'Petrol')
bmw = vehicle(4, 'White', 'Diesel')
```

```python
class railone:

    def __init__(self, train_name,
                  passenger_name, age,
                    gender,seat_no):
        self.train_name = train_name
        self.passenger_name = passenger_name
        self.age = age
        self.gender = gender
        self.seat_no = seat_no

    def traininfo(self):
        print("Train Name: ", self.train_name)

    def passengerinfo(self):
        print("Passenger Name: ", self.passenger_name)
        print("Age: ", self.age)  
        print("Gender: ", self.gender)
        print("Seat Number: ", self.seat_no)

train_name = input("Enter Train Name: ")
passenger_name = input("Enter Passenger Name: ")
age = int(input("Enter Age: "))
gender = input("Enter Gender: ")
seat_no = input("Enter Seat Number: ") 

Maininfo = railone(train_name, passenger_name, age, gender, seat_no)
```

```python
railone.traininfo(Maininfo)
```

```python
railone.passengerinfo(Maininfo)
```
