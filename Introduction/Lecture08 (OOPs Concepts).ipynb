```python
class railone: # class
    irctc = 'Indian Railways' # Class Variable / Class Attribute
    def __init__(self, train_name, passenger_name, age, gender): # Constructor / Special Method / Special Function / Instance Method
        self.train_name = train_name              # Instance Attribute 1
        self.passenger_name = passenger_name      # Instance Attribute 2
        self.age = age                            # Instance Attribute 3
        self.gender = gender                      # Instance Attribute 4

        # Define Class Method
    def passengerinfo(self):
        self.destination = input('Enter Destination: ')    # local attribute / local variable
        print(f'Destination: {self.destination}')
        print(f'Train Name: {self.train_name}')
        print(f'Passenger Name: {self.passenger_name}')
        print(f'Age: {self.age}')
        print(f'Gender: {self.gender }')

# Define Class Method
    def bookinginfo(self):
        self.booking_available = 10  # local attribute
        self.booking = int(input('Enter the booking : '))
        self.fare = 100
        self.choice = input('Enter the booking choice :').lower()

        if self.choice == 'firstclass':

            if (self.booking <= self.booking_available):
                print('Booking is Available')
                self.fare = self.fare * self.booking
                self.booking_available = self.booking_available - self.booking
                print(f'Train Name: {self.train_name}')
                print(f'Passanger Name: {self.passenger_name}')
                print(f'The fare is {self.fare}')
                print(f'Booking Available: {self.booking_available}')
            else:
                print('Booking is Not Available')

train_name = input('Enter Train Name: ')          # Input from User 
passenger_name = input('Enter Passenger Name: ')  # Input from User
age = int(input('Enter Age: '))                   # Input from User
gender = input('Enter Gender: ')                  # Input from User

passenger1 = railone(train_name, passenger_name, age, gender) # Object Creation / Object Instantiation
```
