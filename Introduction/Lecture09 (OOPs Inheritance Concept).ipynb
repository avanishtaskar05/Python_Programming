```python
class employee:    # Define class
    def __init__(self, name, age,
                  department, salary):  # Define constructor
        
        self.name = name                    #Instance Variables 1
        self.age = age                      #Instance Variables 2
        self.department = department        #Instance Variables 3
        self.salary = salary                #Instance Variables 4

    def employee_info(self):                # Define method
        print("Employee Name:", self.name)
        print("Employee Age:", self.age)
        print("Employee Department:", self.department)
        print("Employee Salary:", self.salary)

class manager(employee):
    def __init__(self, name, age, department,
                  salary,  designation):       # Define constructor
        super().__init__(name, age,
                          department, salary)  # Call the constructor of the parent class
        self.designation = designation         # Instance Variable 5

    def manager_info(self):                      # Define method
        self.employee_info()     
        print("Manager Name:", self.name)             # Call the method of the parent class
        print("Manager Designation:", self.designation)

emp1 = employee("Darshan", 30, "IT", 50000)                 # Create object of the class employee
manager1 = manager("Avanish", 40, "IT", 80000, "Team Lead")  # Create object of the class manager
```

```python
emp1.employee_info()
```

```python
manager1.manager_info()
```
