```python
class bank:# Parent class
    # Define Constructor
    def __init__(self, acct_no, acct_holder,
                 age, gender, email_id, address, 
                 mobile_no, IFSC_code, user_id, passcode):
        self.acct_no = acct_no              # Instance Variable 1
        self.acct_holder = acct_holder      # Instance Variable 2
        self.age = age                      # Instance Variable 3
        self.gender = gender                # Instance Variable 4
        self.email_id = email_id            # Instance Variable 5
        self.address = address              # Instance Variable 6
        self.mobile_no = mobile_no          # Instance Variable 7
        self.IFSC_code = IFSC_code          # Instance Variable 8
        self.user_id = user_id              # Instance Variable 9
        self.passcode = passcode            # Instance Variable 10

class saving_acct(bank): # Child class
    # Define Constructor
    def __init__(self, interest_rate, current_balance): 
        super().__init__(acct_no, acct_holder, 
                         age, gender, email_id, 
                         address, mobile_no, IFSC_code,
                         user_id, passcode)
        self.interest_rate = interest_rate      # Instance Variable 10
        self.current_balance = current_balance  # Instance Variable 11
    
```

```python
acct_no = input("Enter your account number: ")
acct_holder = input("Enter your account holder name: ")
age = int(input("Enter your age: "))
gender = input("Enter your gender: ")
email_id = input("Enter your email ID: ")
address = input("Enter your address: ")
mobile_no = input("Enter your mobile number: ")
IFSC_code = input("Enter your IFSC code: ")
user_id = input("Enter your user ID: ")
passcode = input("Enter your passcode: ")
interest_rate = float(input("Enter your interest rate: "))
current_balance = float(input("Enter your current balance: "))  
```

```python
hdfc_1 = saving_acct(interest_rate, current_balance) # Object of child class
```
