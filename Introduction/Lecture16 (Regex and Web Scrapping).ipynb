```python
# regex = Regular Expression for finding patterns from text
# re = regular expression
import re

sentence = "Hello World"
print(sentence)
```

```python
sentence =  "\t\t\t\nHello World"
print(sentence)
```

```python
# r raw string = considers escape characters like \t, \n, ' '
sentence =  r"\t\t\t\nHello World"
print(sentence)
```

```python
import re
text = '''Hello World,I am excited to learn python programming'''
pattern = re.compile(r'a')         # Reading and finding a
matches = pattern.finditer(text)   # finditeration
for i in matches:
    print(i)
```

```python
text = """Customer name: Avanish Taskar.
i am 20 years old.
The phone no is 3243827431.
Order ID: ORD12345.
Delivery expected in 3 days."""
result = re.findall("[0-9]",text)
print(result)
```
